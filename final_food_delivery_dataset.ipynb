```python
# Food Delivery Data Analysis
### Innomatics Research Labs – Advanced GenAI Internship Entrance Test

#This notebook loads data from CSV, JSON, and SQL files,
#merges them into a final dataset, and performs analysis
#to answer the given questions.

import pandas as pd
import sqlite3
```

```python
import numpy as np
```

```python
#Load orders.csv
orders = pd.read_csv("orders.csv")
orders.head()
```

```python
#Load users.json
users = pd.read_json("users.json")
users.head()
```

```python
#Load restaurants.sql
conn = sqlite3.connect("restaurants.db")

# Read the SQL file content
with open("restaurants.sql", "r") as f:
    sql_script = f.read()

# Execute the SQL script to create the table and insert data
conn.executescript(sql_script)

restaurants = pd.read_sql("SELECT * FROM restaurants", conn)
restaurants.head()
```

```python
#Merge Orders + Users (LEFT JOIN)
merged_df = orders.merge(users, on="user_id", how="left")
merged_df.head()
```

```python
#Merge with Restaurants Data (LEFT JOIN)
final_df = merged_df.merge(restaurants, on="restaurant_id", how="left")
final_df.head()
```

```python
#Save Final Dataset (VERY IMPORTANT)
final_df.to_csv("final_food_delivery_dataset.csv", index=False)
```

```python
#Total orders by Gold members
final_df[final_df["membership"] == "Gold"].shape[0]
```

```python
#Revenue by city
final_df.groupby("city")["total_amount"].sum()
```

```python
#Average order value (Gold)
final_df[final_df["membership"] == "Gold"]["total_amount"].mean()
```
